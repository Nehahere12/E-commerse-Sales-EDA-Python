```python
# import pandas
import pandas as pd  
import plotly.express as px
import plotly.graph_objects as go
import plotly.io as pio
pio.templates.default="plotly_white"
```

```python
# to read the file
data =pd.read_csv("Sample - Superstore.csv",encoding='latin-1')
data.head()
```

```python
# to get major details about data
data.describe()
```

```python
# TO GET COLUMN WISE INFORMATION
data.info()
```

# Convert order and shipped date to datetime

```python
data['Order Date']= pd.to_datetime(data['Order Date'])
data['Ship Date']=pd.to_datetime(data['Order Date'])
```

```python
data.info()
```

# To add new  Date-Based Columns

```python
data['Order Month']=data['Order Date'].dt.month
data['Order Year']=data['Order Date'].dt.year
data['Order Day Of Week']=data['Order Date'].dt.dayofweek
```

```python
# Viewed DATA with 3 new columns
data.head()
```

```python
# Monthly Sales Analysis
Monthly_Sales=data.groupby('Order Month')['Sales'].sum().reset_index()
fig= px.line(Monthly_Sales,
             x='Order Month',
             y='Sales',
             title='Monthly Sales Analysis',
            markers=True,  # Adds dots to data points
              color_discrete_sequence=['#636EFA'])  # Sets base line color)

fig.update_traces(line=dict(width=4), marker=dict(size=10))
fig.update_layout(title_font_color="#FF08E8", xaxis_title_font_color="#FF08E8", yaxis_title_font_color="#FF08E8")
fig.show()
```

# **Sales Analysis by Category**

```python
import plotly.express as px
import pandas as pd

# Grouping and summing the data
Category_Sales_Analysis = data.groupby('Category')['Sales'].sum().reset_index()

# Creating the pie chart
fig = px.pie(Category_Sales_Analysis, values='Sales', names='Category', hole=0.5, 
           color_discrete_sequence=px.colors.qualitative.Pastel)

# Updating traces and layout
fig.update_traces(textposition='inside', textinfo='percent+label')
fig.update_layout(title_text='Sales Analysis by Category', title_font=dict(size=24))

# Display the chart
fig.show()
```

# **Sales analysis by sub-category**

```python
SubCategory_Sales_Analysis = data.groupby('Sub-Category')['Sales'].sum().reset_index()

# Creating the colorful bar chart
fig = px.bar(
    SubCategory_Sales_Analysis,
    x='Sub-Category',
    y='Sales',
    color='Sub-Category',
    color_discrete_sequence=px.colors.qualitative.Plotly,
    title='Sales Analysis By Sub-Category',
    width=1000,   # Increased width
    height=600    # Increased height
)

# Customizing fonts, layouts, and sorting order
fig.update_layout(
    title_font=dict(size=22),
    xaxis_title="Sub-Category",
    yaxis_title="Total Sales",
    xaxis={
        'categoryorder': 'total descending',  # Highest sales first
        'tickfont': dict(size=8)              # Decreased font size for x-axis
    }
)

fig.update_traces(width=0.4)

fig.update_layout(
    legend=dict(
        orientation="h",      # Sets layout to horizontal
        yanchor="bottom",     # Anchors the legend from its bottom edge
        y=0.9,               # Places it just above the plot area (1.0 is the top edge)
        xanchor="center",     # Anchors the legend from its horizontal center
        x=0.5,                # Centers it horizontally
        font=dict(size=8)     # Sets the font size to 8
    )
)

# Display the chart
fig.show()
```

# **Sales and profit analysis by customer Segment**

```python
import plotly.graph_objects as go
from plotly import colors

salesNprofit_by_segment = data.groupby('Segment').agg({'Sales':'sum','Profit':'sum'}).reset_index()
color_palette = colors.qualitative.Pastel

# FIXED: Capitalized 'Figure' to avoid a syntax error
fig = go.Figure()

fig.add_trace(go.Bar(
    x=salesNprofit_by_segment['Segment'],
    y=salesNprofit_by_segment['Sales'],
    name='Sales',
    marker_color=color_palette[0]      # Uses the 1st pastel color
))

fig.add_trace(go.Bar(
    x=salesNprofit_by_segment['Segment'],
    y=salesNprofit_by_segment['Profit'],
    name='Profit',
    marker_color=color_palette[1]      # FIXED: Changed index to 1 so Profit has a unique color
))

fig.update_layout(
    title='Sales and profit analysis by customer segment',
    xaxis_title='Customer Segment',
    yaxis_title='Amount',
    # Added horizontal top legend with size 8 font
    legend=dict(
        orientation="h",       # Horizontal layout
        yanchor="bottom",      # Anchor at the bottom of the legend box
        y=1.02,                # Push slightly above the upper plot boundary
        xanchor="center",      # Anchor at the horizontal center of the legend
        x=0.5,                 # Position exactly in the middle of the X-axis
        font=dict(size=8)      # Set font size to 8
    )
)

fig.show()
```

# **Sales to profit ratio By segment**

```python
sales_profit_by_segment=data.groupby('Segment').agg({'Sales':'sum','Profit':'sum'}).reset_index()
sales_profit_by_segment['Sales_To_Profit_Ratio']=sales_profit_by_segment['Sales']/sales_profit_by_segment['Profit']
print(sales_profit_by_segment[['Segment','Sales','Profit','Sales_To_Profit_Ratio']])

# Creating the colorful bar chart
fig = px.bar(
sales_profit_by_segment,
    x='Segment',
    y='Sales_To_Profit_Ratio',
    color='Sales_To_Profit_Ratio',
    color_discrete_sequence=px.colors.qualitative.Plotly,
    title='Sales To Profit Ratio',
    width=1000,   # Increased width
    height=600    # Increased height
)

# Customizing fonts, layouts, and sorting order
fig.update_layout(
    title_font=dict(size=22),
    xaxis_title="Segment",
    yaxis_title="Sales To Profit Ratio",
)

fig.update_traces(width=0.4)

# Display the chart
fig.show()
```
